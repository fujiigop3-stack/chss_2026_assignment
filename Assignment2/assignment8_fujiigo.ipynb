## 課題1

以下の図を再現してください

- 年齢の分布を、異なる階級幅を使って、4つのサブプロットをわけて描画する
   - 階級幅を適当で決めてよい
- 各サブプロットには、
    - x軸ラベル、y軸ラベル、サブプロット番号(a-d)を指定する
    - 水平方向のグリッド線を追加する
    - ヒストグラムの色を`lightblue`に変更する

![](./Figure1.png)

```python
import matplotlib.pyplot as plt


age = df["age"].dropna()

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

bins = [5, 10, 20, 30]
labels = ["(a)", "(b)", "(c)", "(d)"]

for ax, b, lab in zip(axes.ravel(), bins, labels):
    ax.hist(age, bins=b, color="lightblue")
    ax.set_xlabel("Age")
    ax.set_ylabel("Frequency")
    ax.set_title(lab)
    ax.grid(axis="y")

plt.tight_layout()
plt.show()
```

## 課題2

以下の図を再現してください

- 男性乗客と女性乗客の年齢の分布をそれぞれ描画する
- 全乗客の年齢分布を示す灰色ヒストグラムも描画する
- 図のタイトル、x軸ラベル、y軸ラベル、凡例を設定する

```python
import matplotlib.pyplot as plt


male = df[df["sex"] == "male"]["age"].dropna()
female = df[df["sex"] == "female"]["age"].dropna()
all_age = df["age"].dropna()

plt.figure(figsize=(8,6))


plt.hist(all_age, bins=20, color="gray", alpha=0.4, label="All")


plt.hist(male, bins=20, alpha=0.7, label="Male")


plt.hist(female, bins=20, alpha=0.7, label="Female")

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.legend()

plt.show()
```
